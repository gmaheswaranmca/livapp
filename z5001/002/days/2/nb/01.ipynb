# Two Sum

Given an array of integers `nums` and an integer `target`, return the indices of the two numbers such that they add up to `target`.

````python
nums = [2,7,11,15]
target = 9

output = [0,1]
print(nums, target, output, sep='\n')
````

## 1. Brute Force Pattern

### Pattern Used

* Nested Loop
* Compare every pair

### Idea

Check all possible pairs and see whether:

```python
nums[i] + nums[j] == target
```

### Time Complexity

* Time: `O(n²)`
* Space: `O(1)`


````python
# Code 
def two_sum_bruteforce(nums, target):
    n = len(nums)

    for i in range(n):
        for j in range(i + 1, n):

            if nums[i] + nums[j] == target:
                return [i, j]

    return []

nums = [2,7,11,15]
target = 9

print(two_sum_bruteforce(nums, target))
````

# 2. Hashing Pattern (Optimized)

### Pattern Used

* Hash Map / Dictionary
* Lookup Optimization

### Main Idea

Instead of checking every pair:

* Store visited numbers in a dictionary
* For every number:

Need:

```python
target - current_number
```

If already present in dictionary:

* pair found

---

## Visual Understanding

Suppose:

```python
nums = [2,7,11,15]
target = 9
```

Current Number = `7`

Need:

```python
9 - 7 = 2
```

If `2` already exists in hashmap:

```python
{2:0}
```

Then:

```python
2 + 7 = 9
```

Answer found.

---

### Time Complexity

* Time: `O(n)`
* Space: `O(n)`

````python
# Code 
def two_sum(nums, target):
    pairs = {}
    for index, num in enumerate(nums):
        another = target - num 
        if another in pairs:
            return pairs[another], index
        
        pairs[num] = index

nums = [2,7,11,15]
target = 9

print(two_sum(nums, target))
````

# 3. Two Pointer Pattern (When Array is Sorted)

### Pattern Used

* Two Pointer Technique

### Important

Works efficiently only when array is sorted.

---

## Idea

* One pointer at beginning
* One pointer at end

If sum is:

* smaller → move left pointer
* larger → move right pointer

---

## Visualization

Sorted Array:

```python
[2,7,11,15]
```

Pointers:

```python
L = 2
R = 15
```

Sum:

```python
17 > 9
```

Move right pointer left.

---

### Time Complexity

* Time: `O(n)`
* Space: `O(1)`

````python
# Code 
def two_sum(nums, target):
    left = 0
    right = len(nums) - 1
    while left < right:
        s = nums[left] + nums[right]
        if s == target:
            return left, right 
        elif target > s: 
            left = left + 1
        else: 
            right = right - 1

nums = [2,7,11,15]
target = 9

print(two_sum(nums, target))
````
