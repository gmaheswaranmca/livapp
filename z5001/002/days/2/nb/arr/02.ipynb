# Best Time to Buy and Sell Stock

Problem:
Given an array where `prices[i]` is the stock price on day `i`, find the maximum profit from **one buy and one sell**.

You must buy before you sell.

---

# Example

```python
prices = [7,1,5,3,6,4]

Buy at 1
Sell at 6

Profit = 5
```

# Pattern 1 — Brute Force (Nested Loop)

## Idea

Try every possible pair:

* Buy on day `i`
* Sell on day `j`
* where `j > i`

Find maximum profit.

---

## Time Complexity

```text
O(n²)
```

## Space Complexity

```text
O(1)
```

````python
def maxProfit(prices):
    n = len(prices)
    max_profit = 0

    for i in range(n):
        for j in range(i + 1, n):
            profit = prices[j] - prices[i]

            if profit > max_profit:
                max_profit = profit

    return max_profit


prices = [7,1,5,3,6,4]
print(maxProfit(prices))
````

# Pattern 2 — Prefix Minimum Tracking

## Idea

Track:

* minimum price seen so far
* maximum profit possible today

At every day:

```text
profit = current_price - minimum_price_so_far
```

Update answer continuously.

---

## Visualization

```text
Prices: [7,1,5,3,6,4]

Min So Far = 1
Current Price = 6

Profit = 6 - 1 = 5
```

---

## Time Complexity

```text
O(n)
```

## Space Complexity

```text
O(1)
```

````python
def maxProfit(prices):

    min_price = float('inf')
    max_profit = 0

    for price in prices:

        if price < min_price:
            min_price = price

        profit = price - min_price

        if profit > max_profit:
            max_profit = profit

    return max_profit


prices = [7,1,5,3,6,4]
print(maxProfit(prices))
````

# Pattern 3 — Sliding Window / Two Pointer Pattern

## Idea

Use two pointers:

* `left` → buying day
* `right` → selling day

Rules:

* If selling gives profit:

  * update answer
* If current price is smaller:

  * move buy day

---

## Time Complexity

```text
O(n)
```

## Space Complexity

```text
O(1)
```

````python
def maxProfit(prices):

    left = 0
    right = 1

    max_profit = 0

    while right < len(prices):

        # profitable
        if prices[right] > prices[left]:

            profit = prices[right] - prices[left]

            if profit > max_profit:
                max_profit = profit

        else:
            # better buying day found
            left = right

        right += 1

    return max_profit


prices = [7,1,5,3,6,4]
print(maxProfit(prices))
````
